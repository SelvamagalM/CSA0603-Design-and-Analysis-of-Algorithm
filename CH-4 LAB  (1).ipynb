```python
def dice_throw(num_sides, num_dice, target):
    dp = [[0] * (target + 1) for _ in range(num_dice + 1)]
    dp[0][0] = 1

    for dice in range(1, num_dice + 1):
        for total in range(1, target + 1):
            for face in range(1, num_sides + 1):
                if total >= face:
                    dp[dice][total] += dp[dice - 1][total - face]

    return dp[num_dice][target]


print("Test Case 1:")
print("Number of ways to reach sum 7:",
      dice_throw(6, 2, 7))

print("\nTest Case 2:")
print("Number of ways to reach sum 10:",
      dice_throw(4, 3, 10))
```

```python
def assembly_line(a1, a2, t1, t2, e1, e2, x1, x2):
    n = len(a1)

    dp1 = [0] * n
    dp2 = [0] * n

    dp1[0] = e1 + a1[0]
    dp2[0] = e2 + a2[0]

    for i in range(1, n):
        dp1[i] = min(dp1[i - 1] + a1[i],
                     dp2[i - 1] + t2[i - 1] + a1[i])

        dp2[i] = min(dp2[i - 1] + a2[i],
                     dp1[i - 1] + t1[i - 1] + a2[i])

    return min(dp1[n - 1] + x1, dp2[n - 1] + x2)


a1 = [4, 5, 3, 2]
a2 = [2, 10, 1, 4]

t1 = [7, 4, 5]
t2 = [9, 2, 8]

e1 = 10
e2 = 12

x1 = 18
x2 = 7

minimum_time = assembly_line(a1, a2, t1, t2, e1, e2, x1, x2)

print("Minimum time required:", minimum_time)
```

```python
def three_assembly_lines(lines, transfer):
    n = len(lines[0])
    dp = [[0] * n for _ in range(3)]

    for j in range(3):
        dp[j][0] = lines[j][0]

    for i in range(1, n):
        for j in range(3):
            dp[j][i] = min(
                dp[k][i - 1] + transfer[k][j] + lines[j][i]
                for k in range(3)
            )

    return min(dp[j][n - 1] for j in range(3)), dp


lines = [
    [5, 9, 3],
    [6, 8, 4],
    [7, 6, 5]
]

transfer = [
    [0, 2, 3],
    [2, 0, 4],
    [3, 4, 0]
]

minimum_time, dp = three_assembly_lines(lines, transfer)

print("Minimum production time:", minimum_time)
print("\nDP Table:")

for row in dp:
    print(row)
```

```python
def min_path_distance(matrix):
    n = len(matrix)

    dp = [[0] * n for _ in range(n)]

    for i in range(n):
        for j in range(n):
            dp[i][j] = matrix[i][j]

    for k in range(n):
        for i in range(n):
            for j in range(n):
                dp[i][j] = min(dp[i][j],
                               dp[i][k] + dp[k][j])

    minimum = min(
        dp[i][j]
        for i in range(n)
        for j in range(n)
        if i != j
    )

    return minimum, dp


matrix1 = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

matrix2 = [
    [0, 10, 10, 10],
    [10, 0, 10, 10],
    [10, 10, 0, 10],
    [10, 10, 10, 0]
]

matrix3 = [
    [0, 1, 2, 3],
    [1, 0, 4, 5],
    [2, 4, 0, 6],
    [3, 5, 6, 0]
]

for i, matrix in enumerate([matrix1, matrix2, matrix3], 1):
    result, dp = min_path_distance(matrix)
    print("Test Case", i)
    print("Minimum path distance:", result)
    print()
```

```python
from itertools import permutations

cities = ['A', 'B', 'C', 'D', 'E']

distance = {
    ('A', 'B'): 10,
    ('A', 'C'): 15,
    ('A', 'D'): 20,
    ('A', 'E'): 25,
    ('B', 'C'): 35,
    ('B', 'D'): 25,
    ('B', 'E'): 30,
    ('C', 'D'): 30,
    ('C', 'E'): 20,
    ('D', 'E'): 15
}

def get_distance(a, b):
    if (a, b) in distance:
        return distance[(a, b)]
    return distance[(b, a)]

start = 'A'
others = [c for c in cities if c != start]

minimum_distance = float('inf')
best_route = None

for perm in permutations(others):
    route = [start] + list(perm) + [start]

    total = 0

    for i in range(len(route) - 1):
        total += get_distance(route[i], route[i + 1])

    if total < minimum_distance:
        minimum_distance = total
        best_route = route

print("Shortest Route:", " -> ".join(best_route))
print("Total Distance:", minimum_distance)
```

```python
def longest_palindrome(s):
    n = len(s)

    if n == 0:
        return ""

    dp = [[False] * n for _ in range(n)]

    start = 0
    max_length = 1

    for i in range(n):
        dp[i][i] = True

    for length in range(2, n + 1):
        for i in range(n - length + 1):
            j = i + length - 1

            if s[i] == s[j]:
                if length == 2 or dp[i + 1][j - 1]:
                    dp[i][j] = True

                    if length > max_length:
                        start = i
                        max_length = length

    return s[start:start + max_length]


print("Test Case 1:", longest_palindrome("babad"))
print("Test Case 2:", longest_palindrome("cbbd"))
```

```python
def longest_unique_substring(s):
    last_seen = {}
    start = 0
    maximum = 0

    for i, char in enumerate(s):
        if char in last_seen and last_seen[char] >= start:
            start = last_seen[char] + 1

        last_seen[char] = i
        maximum = max(maximum, i - start + 1)

    return maximum


print("Test Case 1:", longest_unique_substring("abcabcbb"))
print("Test Case 2:", longest_unique_substring("bbbbb"))
print("Test Case 3:", longest_unique_substring("pwwkew"))
```

```python
def word_break(s, wordDict):
    n = len(s)

    dp = [False] * (n + 1)
    dp[0] = True

    for i in range(1, n + 1):
        for word in wordDict:
            if i >= len(word) and dp[i - len(word):i][0] if False else False:
                pass

        for j in range(i):
            if dp[j] and s[j:i] in wordDict:
                dp[i] = True
                break

    return dp[n]


print("Test Case 1:",
      word_break("leetcode", ["leet", "code"]))

print("Test Case 2:",
      word_break("applepenapple", ["apple", "pen"]))

print("Test Case 3:",
      word_break("catsandog",
                 ["cats", "dog", "sand", "and", "cat"]))
```

```python
def can_segment(s, dictionary):
    n = len(s)
    dp = [False] * (n + 1)
    dp[0] = True

    for i in range(1, n + 1):
        for j in range(i):
            if dp[j] and s[j:i] in dictionary:
                dp[i] = True
                break

    return dp[n]


dictionary = {
    "i", "like", "sam", "sung", "samsung",
    "mobile", "ice", "cream", "icecream",
    "man", "go", "mango"
}

print("Input: ilike")
print("Output:", "Yes" if can_segment("ilike", dictionary) else "No")

print("\nInput: ilikesamsung")
print("Output:",
      "Yes" if can_segment("ilikesamsung", dictionary) else "No")
```

```python
def full_justify(words, maxWidth):
    result = []
    i = 0

    while i < len(words):
        line_length = len(words[i])
        j = i + 1

        while j < len(words) and line_length + 1 + len(words[j]) <= maxWidth:
            line_length += 1 + len(words[j])
            j += 1

        line_words = words[i:j]
        total_word_length = sum(len(word) for word in line_words)
        spaces = maxWidth - total_word_length

        if j == len(words) or len(line_words) == 1:
            line = " ".join(line_words)
            line += " " * (maxWidth - len(line))
        else:
            gaps = len(line_words) - 1
            space_each = spaces // gaps
            extra = spaces % gaps

            line = ""

            for k in range(gaps):
                line += line_words[k]
                line += " " * (space_each + (1 if k < extra else 0))

            line += line_words[-1]

        result.append(line)
        i = j

    return result


words = [
    "This", "is", "an", "example",
    "of", "text", "justification."
]

output = full_justify(words, 16)

for line in output:
    print('"' + line + '"')
```

```python
class WordFilter:

    def __init__(self, words):
        self.words = words

    def f(self, pref, suff):
        answer = -1

        for i in range(len(self.words)):
            if self.words[i].startswith(pref) and self.words[i].endswith(suff):
                answer = i

        return answer


words = ["apple"]

wordFilter = WordFilter(words)

print("Output:", wordFilter.f("a", "e"))
```

```python
INF = float('inf')

routers = ['A', 'B', 'C', 'D', 'E', 'F']

edges = [
    ('A', 'B', 1),
    ('A', 'C', 5),
    ('B', 'C', 2),
    ('B', 'D', 1),
    ('C', 'E', 3),
    ('D', 'E', 1),
    ('D', 'F', 6),
    ('E', 'F', 2)
]

def floyd(matrix):
    n = len(matrix)

    dist = [row[:] for row in matrix]

    for k in range(n):
        for i in range(n):
            for j in range(n):
                dist[i][j] = min(
                    dist[i][j],
                    dist[i][k] + dist[k][j]
                )

    return dist


def create_matrix():
    matrix = [[INF] * 6 for _ in range(6)]

    for i in range(6):
        matrix[i][i] = 0

    for a, b, w in edges:
        i = routers.index(a)
        j = routers.index(b)
        matrix[i][j] = w
        matrix[j][i] = w

    return matrix


matrix = create_matrix()
dist = floyd(matrix)

print("Before link failure:")
print("Router A to Router F =", dist[0][5])

matrix[1][3] = INF
matrix[3][1] = INF

dist = floyd(matrix)

print("\nAfter link failure:")
print("Router A to Router F =", dist[0][5])
```

```python
INF = float('inf')

def floyd_warshall(n, edges):
    dist = [[INF] * n for _ in range(n)]

    for i in range(n):
        dist[i][i] = 0

    for u, v, w in edges:
        dist[u][v] = w

    for k in range(n):
        for i in range(n):
            for j in range(n):
                dist[i][j] = min(
                    dist[i][j],
                    dist[i][k] + dist[k][j]
                )

    return dist


edges1 = [
    (1, 0, 2),
    (0, 2, 3),
    (2, 3, 1),
    (3, 0, 6),
    (2, 1, 7)
]

dist = floyd_warshall(4, edges1)

print("Shortest path from C to A:", dist[2][0])
```

```python
def optimal_bst(keys, freq):
    n = len(keys)

    cost = [[0] * n for _ in range(n)]
    root = [[0] * n for _ in range(n)]

    for i in range(n):
        cost[i][i] = freq[i]
        root[i][i] = i

    for length in range(2, n + 1):
        for i in range(n - length + 1):
            j = i + length - 1

            total_freq = sum(freq[i:j + 1])
            cost[i][j] = float('inf')

            for r in range(i, j + 1):
                left = cost[i][r - 1] if r > i else 0
                right = cost[r + 1][j] if r < j else 0

                current = left + right + total_freq

                if current < cost[i][j]:
                    cost[i][j] = current
                    root[i][j] = r

    return cost, root


keys = ['A', 'B', 'C', 'D']
freq = [0.1, 0.2, 0.4, 0.3]

cost, root = optimal_bst(keys, freq)

print("Cost:", cost[0][3])

print("\nCost Matrix:")
for row in cost:
    print(row)

print("\nRoot Matrix:")
for row in root:
    print(row)
```

```python
from functools import lru_cache

def cat_mouse_game(graph):

    @lru_cache(None)
    def solve(mouse, cat, turn, depth):
        if mouse == 0:
            return 1

        if mouse == cat:
            return 2

        if depth >= 2 * len(graph):
            return 0

        if turn == 0:
            for next_node in graph[mouse]:
                result = solve(next_node, cat, 1, depth + 1)

                if result == 1:
                    return 1

            return 0

        else:
            for next_node in graph[cat]:
                if next_node == 0:
                    continue

                result = solve(mouse, next_node, 0, depth + 1)

                if result == 2:
                    return 2

            return 0

    return solve(1, 2, 0, 0)


graph1 = [
    [2, 5],
    [3],
    [0, 4, 5],
    [1, 4, 5],
    [2, 3],
    [0, 2, 3]
]

graph2 = [
    [1, 3],
    [0],
    [3],
    [0, 2]
]

print("Test Case 1:", cat_mouse_game(graph1))
print("Test Case 2:", cat_mouse_game(graph2))
```

```python
import heapq

def max_probability(n, edges, probabilities, start, end):

    graph = [[] for _ in range(n)]

    for i in range(len(edges)):
        u, v = edges[i]
        p = probabilities[i]

        graph[u].append((v, p))
        graph[v].append((u, p))

    best = [0.0] * n
    best[start] = 1.0

    heap = [(-1.0, start)]

    while heap:
        negative_prob, node = heapq.heappop(heap)
        probability = -negative_prob

        if node == end:
            return probability

        for next_node, edge_prob in graph[node]:
            new_prob = probability * edge_prob

            if new_prob > best[next_node]:
                best[next_node] = new_prob
                heapq.heappush(heap, (-new_prob, next_node))

    return 0.0


edges = [
    [0, 1],
    [1, 2],
    [0, 2]
]

probabilities = [0.5, 0.5, 0.2]

print("Maximum probability:",
      max_probability(3, edges, probabilities, 0, 2))
```

```python
def unique_paths(m, n):
    dp = [[0] * n for _ in range(m)]

    for i in range(m):
        dp[i][0] = 1

    for j in range(n):
        dp[0][j] = 1

    for i in range(1, m):
        for j in range(1, n):
            dp[i][j] = dp[i - 1][j] + dp[i][j - 1]

    return dp[m - 1][n - 1]


print("Test Case 1:")
print("Number of unique paths:", unique_paths(3, 7))

print("\nTest Case 2:")
print("Number of unique paths:", unique_paths(3, 2))
```

```python
def good_pairs(nums):
    count = {}

    pairs = 0

    for num in nums:
        if num in count:
            pairs += count[num]

        count[num] = count.get(num, 0) + 1

    return pairs


print("Test Case 1:")
print("Number of good pairs:",
      good_pairs([1, 2, 3, 1, 1, 3]))

print("\nTest Case 2:")
print("Number of good pairs:",
      good_pairs([1, 1, 1, 1]))
```

```python
import heapq

def network_delay_time(times, n, k):

    graph = [[] for _ in range(n + 1)]

    for u, v, w in times:
        graph[u].append((v, w))

    distance = [float('inf')] * (n + 1)
    distance[k] = 0

    heap = [(0, k)]

    while heap:
        current_time, node = heapq.heappop(heap)

        if current_time > distance[node]:
            continue

        for next_node, weight in graph[node]:

            new_time = current_time + weight

            if new_time < distance[next_node]:
                distance[next_node] = new_time
                heapq.heappush(heap, (new_time, next_node))

    maximum = max(distance[1:])

    if maximum == float('inf'):
        return -1

    return maximum


print("Test Case 1:")
print("Network delay time:",
      network_delay_time(
          [[2, 1, 1], [2, 3, 1], [3, 4, 1]],
          4,
          2
      ))

print("\nTest Case 2:")
print("Network delay time:",
      network_delay_time(
          [[1, 2, 1]],
          2,
          1
      ))

print("\nTest Case 3:")
print("Network delay time:",
      network_delay_time(
          [[1, 2, 1]],
          2,
          2
      ))
```
